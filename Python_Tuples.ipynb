```python
#this tuple code using a real-time example from a shopping application.

cart = ("Mobil", "Loptap", "Headphones")
New_Item = ("Smart Watch",)
cart = cart + New_Item
print(cart)
```

```python
cart = ("Mobil", "Loptap", "Headphones")
New_Item = ("Smart Watch",)
cart += New_Item
print(cart)
```

```python
cart = ('Mobil', 'Loptap', 'Headphones', 'Smart Watch')

y = list(cart)

y.remove("Headphones")

cart = tuple(y)

print(cart)
```

Using Asterisk*

```python
cart = ('Mobil', 'Loptap', 'Headphones', 'Smart Watch')
(Rose, Jasmine, Lily, Sunflower) = cart
print(Rose)
print(Jasmine)
print(Lily)
print(Sunflower)
```

```python
cart = ('Mobil', 'Loptap', 'Headphones', 'Smart Watch')
(Rose, Jasmine, *Lily) = cart

print(Rose)
print(Jasmine)
print(Lily)
```

```python
cart = ('Mobil', 'Loptap', 'Headphones', 'Smart Watch')
(Rose, *Jasmine, Lily) = cart
print(Rose)
print(Jasmine)
print(Lily)
```

Python - Loop Tuples

```python
cart = ("Mobil", "Loptap", "Headphones", "Smart Watch")

for i in range(len(cart)):
    print(cart[i])

```

```python
flower = ("Rose", "Jasmine", "Lily", "Lotus", "Sunflower")

for i in range(len(flower)):
  print(flower[i])
```

```python
flower = ("Rose", "Jasmine", "Lily", "Lotus", "Sunflower")

for x in flower:
  print(x)
```
