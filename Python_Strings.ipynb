```python
#Check String

txt = "The best things in life are free!"
print("free" in txt)
```

```python
txt = "The best things in life are free!"
if "free" in txt:
  print("Yes, 'free' is present.")
```

```python
#Check if NOT
txt = "The best things in life are free!"
print("expensive" not in txt)
```

```python
txt = "The best things in life are free!"
if "expensive" not in txt:
  print("No, 'expensive' is NOT present.")
```

Python - Slicing Strings

```python
#Slicing
b = "Hello, World!"
print(b[2:5])
```

```python
#Slice From the Start

A = "Hello, World!"
print(A[:5])
```

```python
#Slice To the End
X = "Hello, World!"
print(b[2:])
```

```python
#Negative Indexing

b = "Hello, World!"
print(b[-5:-2])
```
