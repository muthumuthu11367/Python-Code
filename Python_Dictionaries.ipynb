Python - Remove Dictionary Items


```python
Item = {
    "Car_Name":"TATA",
    "Model":"Safai",
    "Year": 2026,
    "price": 10000000
}
Item.pop("Year")
print(Item)
```

```python
# method removes the last inserted item (in versions before 3.7, a random item is removed instead

Item = {
    "Car_Name":"TATA",
    "Model":"Safai",
    "Year": 2026,
    "price": 10000000
}
Item.popitem()
print(Item)
```

```python
#The del keyword removes the item with the specified key name
Item = {
    "Car_Name":"TATA",
    "Model":"Safai",
    "Year": 2026,
    "price": 10000000
}
del Item["Year"]
print(Item)
```

```python
#The del keyword can also delete the dictionary completely

item = {
    "Car_Name": "TATA",
    "Model": "Safai",
    "Year": 2026,
    "price": 10000000
}
del item
print(item)
```

```python
item = {
    "Car_Name": "TATA",
    "Model": "Safai",
    "Year": 2026,
    "price": 10000000
}
item.clear()
print(item)
```

Loop Through a Dictionary


```python
item = {
    "Car_Name": "TATA",
    "Model": "Safai",
    "Year": 2026,
    "price": 10000000
}

for x in item:
  print(x)
```

```python
item = {
    "Car_Name": "TATA",
    "Model": "Safai",
    "Year": 2026,
    "price": 10000000
}

for i in item:
  print(item[i])
```

```python
item = {
    "Car_Name": "TATA",
    "Model": "Safai",
    "Year": 2026,
    "price": 10000000
}

for x in item.values():
  print(x)
```

```python
item = {
    "Car_Name": "TATA",
    "Model": "Safai",
    "Year": 2026,
    "price": 10000000
}
for x in item.keys():
  print(x)
```

```python
item = {
    "Car_Name": "TATA",
    "Model": "Safai",
    "Year": 2026,
    "price": 10000000
}
for y,x in item.items():
  print(y,x)
```

Python - Copy Dictionaries


```python
item = {
    "Car_Name": "TATA",
    "Model": "Safai",
    "Year": 2026,
    "price": 10000000
}

my = item.copy()
print(my)
```

```python
item = {
    "Car_Name": "TATA",
    "Model": "Safai",
    "Year": 2026,
    "price": 10000000
}

my_item = dict(item)
print(my_item)
```

Python - Nested Dictionaries


```python
Name1 = {
    "Car_Name":{
        "Name": "Tata",
        "Model": "Safai",
        "Year": 2026
    },
    "Car_Name":{
        "Name": "Tata",
        "Model": "Safai",
        "Year": 2026
    },
    "Car_Name":{
        "Name": "Tata",
        "Model": "Safai",
        "Year": 2026
    }


}

print(Name1)
```

```python
Student1 = {
    "Name":"Arun",
    "Age": 20,
    "DOB": 20-10-2006

}
Students2 ={
    "Name":"Muthu",
    "Age": 25,
    "DOB": 18-10-2001
}
Students3 = {
    "Name": "Amaresh",
    "Age":22,
    "DOB": 18-10-2004

}
Students4 = {
    "Name": "Thiyagesh",
    "Age" : 23,
    "DOB": 18-10-2003

}

My_Students = {
    "Student1": Student1,
    "Student2": Students2,
    "Student3": Students3,
    "Student4": Students4
}

print(My_Students)
```

```python
myfamily = {
  "child1" : {
    "name" : "Emil",
    "year" : 2004
  },
  "child2" : {
    "name" : "Tobias",
    "year" : 2007
  },
  "child3" : {
    "name" : "Linus",
    "year" : 2011
  }
}

for x, obj in myfamily.items():
    print(x)

    for y in obj:
        print(y + ':', obj[y])
```
