Remove List Items

```python
#The remove() method removes the specified item
Flowers_Items_ = ["Rose", "Lity", "Merigold", "Lotus", "Primrose", "Hoibiscus"]
Flowers_Items_.remove("Rose")
print(Flowers_Items_)
```

```python
#more than one item with the specified value, the remove() method removes the first occurrence
Animals_Name_List = ["Cow", "Dog", "Cat", "Goat", "Sheep", "Hours", "Dog"]
Animals_Name_List.remove("Dog")
print(Animals_Name_List)
```

```python
# do not specify the index, the pop() method removes the last item
Flowers_Items_ = ["Rose", "Lity", "Merigold", "Lotus", "Primrose", "Hoibiscus"]
Flowers_Items_.pop()
print(Flowers_Items_)
```

List Comprehension

```python
Flowers = ["Lity", "Rose", "Lotus", "merigold"]
newlist = []
for x in Flowers:
  if "o" in x:
    newlist.append(x)
print(newlist)
```

```python
Flower=["Lity", "Rose", "Lotus", "merigold"]
newlist =[x for x in Flower if "o" in x]
print(newlist)
```

Sort List Alphanumerically

```python
#You can also customize your own function by using the keyword argument key = function

#The function will return a number that will be used to sort the list (the lowest number first

def myfunc(n):
  return abs(n - 100)

number = [100, 110, 190, 105, 250, 75,]

number.sort(key=myfunc)

print(number)
```

```python
#a sort() method that will sort the list alphanumerically, ascending, by default

item = ["Lity", "Rose", "Lotus", "Merigold"]
item.sort()
print(item)
```
